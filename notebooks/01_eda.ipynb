```python
data['mo_sin_old_il_acct_was_missing'] = data['mo_sin_old_il_acct'].isna().astype(int)
data['mo_sin_old_il_acct'] = data['mo_sin_old_il_acct'].fillna(data['mo_sin_old_il_acct'].max())
```

```python
fill_with_max = ['mths_since_recent_bc', 'mths_since_recent_inq']
fill_with_zero = ['bc_open_to_buy', 'bc_util', 'percent_bc_gt_75', 'num_tl_120dpd_2m']
for i in fill_with_max:
    data[i + "_was missing"] = data[i].isna().astype(int)
    data[i] = data[i].fillna(data[i].max())
for j in fill_with_zero:
    data[j+ "_was missing"] = data[j].isna().astype(int)
    data[j] = data[j].fillna(0)
```

```python
data['emp_length_was_missing'] = data['emp_length'].isna().astype(int)
data['emp_length'] = data['emp_length'].fillna(0)
```
